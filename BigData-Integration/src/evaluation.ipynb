Evaluation of results obtained from llm responses

```python
import numpy as np
import matplotlib.pyplot as plt
import paths
from evaluation import table_match_eval, att_table_match_eval
```

Source Tables evaluation

```python
filename = paths.LLM_RESPONSE + "match_tables.txt"
precision, recall, f1_score = table_match_eval(filename, needResults=True)

values = [precision, recall, f1_score]

plt.figure(figsize=(6, 4))
bars = plt.bar(['Precision', 'Recall', 'F1-score'], values, color=['blue', 'orange', 'green'])

# Aggiungi i valori numerici sopra le barre
for bar in bars:
    yval = bar.get_height()  # Ottieni l'altezza della barra (cioè il valore)
    plt.text(bar.get_x() + bar.get_width() / 2, yval + 0.02, f'{yval:.2f}', 
             ha='center', va='bottom', fontsize=10)  # Posiziona il testo sopra la barra

plt.xlabel('Metrics')
plt.ylabel('Score')
plt.title('Precision, Recall, and F1-score Comparison')
plt.ylim(0, 1)  # Scores are between 0 and 1
plt.show()
```

Source tables and attributes evaluation

```python
filename = paths.LLM_RESPONSE + "att_match_tables.txt"
precision, recall, f1_score, att_precision, att_recall, att_f1_score = att_table_match_eval(filename, needResults=True)

metrics = ['Precision', 'Recall', 'F1-score']

fig, axes = plt.subplots(1, 2, figsize=(8, 4))
# 1 plot
values = [precision, recall, f1_score]
axes[0].bar(metrics, values, color=['blue', 'orange', 'green'])
axes[0].set_xlabel('Metrics')
axes[0].set_ylabel('Score')
axes[0].set_title('Tables evaluation')
axes[0].set_ylim(0, 1)

for i, v in enumerate(values):
    axes[0].text(i, v + 0.02, f'{v:.2f}', ha='center', va='bottom', fontsize=10)

# 2 plot
values = [att_precision, att_recall, att_f1_score]
axes[1].bar(metrics, values, color=['blue', 'orange', 'green'])
axes[1].set_xlabel('Metrics')
axes[1].set_ylabel('Score')
axes[1].set_title('Attributes evaluation')
axes[1].set_ylim(0, 1)


for i, v in enumerate(values):
    axes[1].text(i, v + 0.02, f'{v:.2f}', ha='center', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()
```
